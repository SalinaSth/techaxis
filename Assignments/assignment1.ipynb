### Task 2

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

if num2 == 0:
    print("For Division and modulus num2 cannot be 0")
else:
    add = num1 + num2
    sub = num1 - num2
    mul = num1 * num2
    div = num1 / num2
    mod = num1 % num2
    exp = num1 ** num2
    f_div = num1 // num2

    print(f'Add: {add}, Subtract: {sub}, Multiply: {mul}, Division: {div}')
    print(f'Modulus: {mod}, Exponentiation: {exp}, Floor Division: {f_div}')  
```
