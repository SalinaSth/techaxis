Task 3: Implement a function that flattens a nested list (e.g., `[[1,2], [3,4], [5,6]] →
[1,2,3,4,5,6]`).


```python
def flatten(new_list):
    flatList = []
    for sublist in new_list:
        flatList.extend(sublist)
        # print('test---', subList)
    return flatList

nested_list = [[1, 2, 3], [6, 7], [5, 66]]
print(flatten(nested_list))
```

### Topic 2 :Tuples & Sets

Task 1: Write a Python program to calculate the sum of all unique elements in a
tuple of numbers

```python
def uniqueTupleSum(num):
    count = 0
    for ele in num:
        if num.count(ele) == 1:
            count += ele
    print('Sum of unique tuple values: ',count)
    return

new_tuple= (1, 2, 3 ,5 ,3 ,8, 2, 4)
uniqueTupleSum(new_tuple)
```

Task 2: Implement a function that takes two sets as input and returns their
intersection, union, and difference.

```python
def setCalc(set1, set2):
    set_inter = set1.intersection(set2)
    set_union = set1.union(set2)
    set_diff = set1.difference(set2)
    print(f'Set intersection: {set_inter}\n',
          f'Set Union: {set_union}\n', 
          f'Set Difference: {set_diff}')
    return 

set1 = {1, 2, 3, 4, 5, 7}
set2 = {2, 3, 4, 7, 8, 9}
setCalc(set1, set2)
```

Task 5: Given a tuple of words, write a function to return the longest word and
its length.


```python
def tupleWordLength(tup_words):
    max_len = 0
    max_word=''
    for i in tup_words:
        length = len(i)
        if length > max_len:
            max_len = length
            max_word = i
    print(f'Longest word is {max_word} and its length is {max_len}')
    return

tup_words = ('apple', 'Joseph', 'Elizabeth', 'cat', 'purple')
tupleWordLength(tup_words) 
```

Task 2. Implement a function that merges two dictionaries and updates common
keys with their sum.

```python
d1 ={'a': 1, 'b': 2, 'c': 3}
d2 ={'c': 2, 'e': 2, 'a': 3}

def merge_sum(d1, d2):
    for k in d2:
        if k in d1:
            d1[k] += d2[k]
        else:
            d1[k] = d2[k] 
    
    print(d1)
    return
merge_sum(d1, d2)
```

finds the three most expensive products.

```python
products = {'Iphone': 1200, 'Laptop': 900, 'Airpods': 800, 'Chair': 500, 'Fan': 1500, 'Watch': 1000}
def expensiveProducts(products):
    sorted_products = sorted(products.items(), key=lambda x:x[1], reverse=True)
    three_exp_products = dict(sorted_products[:3])

    return three_exp_products
expensiveProducts(products)
```

### Topic 4 :String Manipulation & Regular Expressions

Task 1. Write a Python function that removes all vowels from a given string.

```python
vowels = 'aeiouAEIOU'
def removeVowels(text):
    for i in vowels:
        text = text.replace(i, "")
    return text

text = 'A lazy fox in jungle.'
removeVowels(text)
```

Task 2. Given a multiline string, write a Python script to extract all valid phone
numbers using regex.

```python
import re

text = """
John: 987-654-3210
Rose: (123) 456-7890
Irene: 123-456-7890
Charlie: 1234567890
"""
pattern = r"\d{3}-\d{3}-\d{4}"
valid_numbers = re.findall(pattern, text)
print(valid_numbers)
```




Task 3. Implement a function that counts the occurrences of each digit (0-9) in a
given string.



```python
text = "My number is 9878543220. Call me at 124-456-7891!"
def countDigits(text):
    digit_count={}
    for char in text:
        if char.isdigit():
            digit_count[char] = text.count(char)
    sorted_digits = dict(sorted(digit_count.items()))
    return sorted_digits
countDigits(text)
```


Task 4. Implement a function that reads a large file in chunks and prints only the
lines that contain a given keyword.

```python
import csv

def search_in_csv(file_path, keyword):
    with open(file_path, 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            if keyword in ','.join(row):
                print(row)

csv_file = "../data/sample.csv"
search_in_csv(csv_file, "Alice")
```

Task 3. Write a Python program that uses the `datetime` module to print the
current date, time, and the number of days left in the year.

```python
from datetime import datetime, date

current_datetime = datetime.now()

print("Current Date:", current_datetime.strftime("%Y-%m-%d"))  # YYYY-MM-DD format
print("Current Time:", current_datetime.strftime("%H:%M:%S"))  # HH:MM:SS format

end_of_year = date(current_datetime.year, 12, 31)
days_left = (end_of_year - current_datetime.date()).days 

print(f"Days left in {current_datetime.year}: {days_left}")
```
