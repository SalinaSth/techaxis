
Task 9. Write a function that computes the cosine similarity between two NumPy
vectors using broadcasting and vectorized operations.
 

```python
def cosSimilarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm1 = np.sqrt(np.sum(vec1 ** 2))  
    norm2 = np.sqrt(np.sum(vec2 ** 2))
    cosineSim = dot_product / (norm1 * norm2) 
    return cosineSim

vec1 = np.arange(5)
vec2 = np.arange(7, 12) 
cosSimilarity(vec1, vec2)
```

Task 2. Write a function that normalizes a given NumPy array using mean and
standard deviation.

```python
def normalizedArr(arr):
    arr_mean = np.mean(arr)
    arr_std = np.std(arr)
    normalized_arr = (arr - arr_mean) / arr_std
    return normalized_arr
ra6 = np.arange(1, 20, 2)
normalizedArr(ra6)
print(f'Original Array:\n{ra6}')
print(f'Normalized Array:\n{normalizedArr(ra6)}')
```

### Topic 5 :Working with Missing Data in NumPy
Task 1. Create a NumPy array with some missing values (`np.nan`). Write a
function that identifies the indices of missing values.

```python
single_arr = np.array([2, 4, 7, 12, np.nan, 23, 6])

def findNanIndex(arr):
    nan_indices = np.where(np.isnan(arr))
    return nan_indices

findNanIndex(single_arr)
```


Task 1. Given an array with missing values, replace all `np.nan` values with the
mean of the non-missing values.

```python
arr_nan = np.array([
    [1, 2, np.nan],
    [4, np.nan, 6],
    [7, 8, 9]
])
mean_nan = np.nanmean(arr_nan)
arr_nan_value = np.where(np.isnan(arr_nan), mean_nan, arr_nan)
print(f'Array with NaN values:\n{arr_nan}\n',
      f'Mean of NaN values:\n{mean_nan}\n',
      f'Array with NaN replaced by mean:\n{arr_nan_value}') 
```


Task 2. Implement a function that removes all rows containing `np.nan` from a
given 2D NumPy array.

```python
np.random.seed(0)
random_array = np.random.randint(1, 10, size=(5, 4)).astype(float)
random_array[1, 2] = np.nan
random_array[3, 0] = np.nan
random_array[4, 3] = np.nan

print(random_array)
def removeNanRows(arr_row):
    arr_row = arr_row[~np.isnan(arr_row).any(axis=1)]
    return arr_row
removed_rows = removeNanRows(random_array)
print(f'Array with NaN rows removed:\n{removed_rows}')
```
