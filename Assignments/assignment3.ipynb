
Task 9. Write a function that computes the cosine similarity between two NumPy
vectors using broadcasting and vectorized operations.
 

```python
def cosSimilarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm1 = np.sqrt(np.sum(vec1 ** 2))  
    norm2 = np.sqrt(np.sum(vec2 ** 2))
    cosineSim = dot_product / (norm1 * norm2) 
    return cosineSim

vec1 = np.arange(5)
vec2 = np.arange(7, 12) 
cosSimilarity(vec1, vec2)
```

standard deviation.

```python
def normalizedArr(arr):
    arr_mean = np.mean(arr)
    arr_std = np.std(arr)
    normalized_arr = (arr - arr_mean) / arr_std
    return normalized_arr
ra6 = np.arange(1, 20, 2)
normalizedArr(ra6)
print(f'Original Array:\n{ra6}')
print(f'Normalized Array:\n{normalizedArr(ra6)}')
```

Task 6. Write a function that detects outliers in a NumPy array using the
Interquartile Range (IQR) method.

```python
def detectOutliers(r_arr4):
    q1 = np.percentile(r_arr4, 25)
    q3 = np.percentile(r_arr4, 75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers_indices = np.where(
        (r_arr4 < lower_bound) | (r_arr4 > upper_bound)
    )
    outlier_vals = r_arr4[outliers_indices]
    print(f'Outlier values: {outlier_vals}')
    return outliers_indices

r_arr4 = np.array([12, 13, 14, 16, 190, 30, 50, 18, 200, 19, 20])
print(f'Original Array:\n{r_arr4}')
detectOutliers(r_arr4)
```

Task 6. Implement Singular Value Decomposition (SVD) on a given matrix.

```python
svd_mat = np.array([
    [3, 1, 1],
    [-1, 3, 1]
])

U, S, VT = np.linalg.svd(svd_mat)

print("Matrix A:\n", svd_mat)
print("\nU (Left singular vectors):\n", U)
print("\nS (Singular values):\n", S)
print("\nV^T (Right singular vectors - transposed):\n", VT)
```
