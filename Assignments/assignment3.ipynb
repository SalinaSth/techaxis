Task 1: Create a NumPy array of shape (6,6) filled with random integers. Extract:
- The second row
- The last two columns
- The diagonal elements

```python
np.random.seed(1)
arr5 = np.random.randint(1, 40, size=(6, 6))
print(arr5)
second_row = arr5[1]
print('Second Row\n',second_row )
last_two_cols = arr5[:, 4:]
print('Last Two Columns\n', last_two_cols)
diagonal_eles = np.diagonal(arr5)
print('Diagonal elements:\n', diagonal_eles)
```


Task 4:. Write a function that takes a NumPy array and reverses its elements
without using loops.

```python
def reverseArr(np_arr):
    reversed_arr = np_arr[::-1]
    return reversed_arr

np_arr = np.arange((5))
reverseArr(np_arr)
```

Task 6:. Create a NumPy array of shape (3, 6) and reshape it into a 2×9 matrix.
Explain how reshaping affects memory allocation.

```python
np_arr2 = np.arange(18).reshape(3, 6)
reshaped_arr = np_arr2.reshape(2, 9)
print(f'Original Array (3, 6):\n {np_arr2} \n',
      f'Reshaped Array (2, 9):\n {reshaped_arr}')
# Reshaping does not create new memory and just changes how the data is viewed 
# The data remains in the same place in memory but the rows and columns(shape)changes.
# So reshaping only alters the structure and not the actual data
```



Task 8:. Create a function that takes a 2D array and returns its transposed
version.


```python
def transposeArr(arr9):
    transposed_arr = np.transpose(arr9)
    print(f'Original Array:\n {arr9} \n')
    print(f'Transposed Array:\n {transposed_arr}')
    return

arr9 = np.arange(12).reshape(3, 4)
transposeArr(arr9)
```

### Topic 3 :Broadcasting & Vectorized Operations
Task 1: Create two NumPy arrays of shape (3,3) and add them using
broadcasting without using loops.

```python
a1 = np.random.randint(1, 20, size=(3, 3))
a2 = np.random.randint(3, 18, size=(3, 3))
print(f'First Array:\n{a1}\n', f'Second Array:\n{a2}')
sum_arr = a1 + a2
print(f'Sum of two arrays:\n{sum_arr}')
```

Task 2. Given a NumPy array, multiply each row by a different scalar using
broadcasting.

```python
np.random.seed(1)
a3 = np.random.randint(1, 15, size=(3, 3))
multiply_arr = a3 *  np.array([3, 4, 5])[:, None] # converts the 1D array [3, 4, 5] into a column vector (shape (3, 1))
print(f'Original Array:\n{a3}\n',
      f'Multiplied Array:\n{multiply_arr}')
```


Task 3. Create a 3×3 matrix and add a 1D array of size 3 to it using broadcasting.

```python
arr_mat = np.arange(9).reshape(3, 3)
print(f'3x3 matrix: \n{arr_mat}')
arr_1d = np.arange(1,4)
print(f'Size 3 array: \n{arr_1d}')
add_arr = arr_mat + arr_1d
print(f'Addition: \n{add_arr}')
```


Task 4. Implement a function that performs element-wise multiplication between
two numpy arrays of different shapes using broadcasting.

```python
def multiplyArr(arr_shape1, arr_shape2):
    multiply_arr = arr_shape1 * arr_shape2
    print(f'First Array:\n{arr_shape1}\n',
          f'Second Array:\n{arr_shape2}')
    print(f'Multiplication:\n {multiply_arr}')
    return 

arr_shape1 = np.arange(12).reshape(4, 3)
arr_shape2 = np.arange(1, 4)
multiplyArr(arr_shape1, arr_shape2)
```


Task 5. Write a Python program to compute the dot product of two matrices

```python
dot_prod = np.dot(arr_shape1, arr_shape2)
dot_prod
```
