Task 2. Given a Series of student scores, write a function to find the mean,
maximum, and minimum scores.

```python
student_score = [91, 20, 40, 45, 70, 95, 80, 87, 33, 65]
# scores = pd.Series(student_score)
# scores
def grade(student_score):
    scores = pd.Series(student_score)
    mean_score = scores.mean()
    max_score = scores.max()
    min_score = scores.min()
    return mean_score, max_score, min_score

mean, maximum, minimum = grade(student_score)
print(f"Mean: {mean}, Max: {maximum}, Min: {minimum}")
```

Task 10. Convert a Pandas Series with categorical values into numerical labels
using `.astype()` or `.factorize()`.

```python
colors = pd.Series(['Red', 'Green', 'Blue', 'Green', 'Red', 'Blue'])
print(f'Colors Series:\n{colors}')
# Convert to categorical data type
numeric_labels = colors.astype('category').cat.codes
print(f'Numeric Labels:\n{numeric_labels}')

# convert to numerical labels using factorize
labels, uniques = pd.factorize(colors)
print("Numerical Labels:", labels)
print("Unique Categories:", uniques)
```

Task 12. Write a function that takes a DataFrame as input and returns the total

```python
def missingValues(df_data):
    total = df_data.isnull().sum()
    return total

data = {
    'Name': ['Alice', 'Bob', None, None, 'Eva'],
    'Age': [25, None, 35, 60, 45],
    'City': ['Kathmandu', 'Lalitpur', 'Bhaktapur', 'Karnali', 'Pokhara'],
}
df_data = pd.DataFrame(data)

missing_counts = missingValues(df_data)
print(missing_counts)
```

Task 11. Detect and replace outliers in a dataset using the IQR method.

```python
Q1 = missing_df2['Age'].quantile(0.25)
Q3 = missing_df2['Age'].quantile(0.75)
IQR = Q3 - Q1

outlier_condition = (missing_df2['Age'] < (Q1 - 1.5 * IQR)) | (missing_df2['Age'] > (Q3 + 1.5 * IQR))
df_no_outliers = missing_df2[~outlier_condition]
print(df_no_outliers)
```
