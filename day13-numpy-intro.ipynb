### Python's `all()` and `any()`

```python
# all and any return boolean
# if no elements = false, if 1 or many then true
if {'a' : 2}:
    print("inside if statement")
else: 
    print("inside else statement")
```

```python
l = [True, [], 4]
```

```python
all(l), any(l)
```

```python
# if string has even space then true return
if -1:
    print("inside if statement")
else: 
    print("inside else statement")
```

```python
np.all(arr2), np.any(arr2)
```
