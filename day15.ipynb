### Broadcasting

```python
import numpy as np
```

```python
arr1= np.arange(9).reshape((3, 3))
arr2= np.arange(3)
print(arr1)
print()
print(arr2)
```

```python
# should match the dimension of array to do any operations
#if arr2 has np.arange(4), it will show valueError
arr2 + arr1
```

```python
arr1 = np.arange(9).reshape((3, 3))
arr1
```

```python
arr1 + 100 # scalar addition, adds to every element by 100
```

```python
arr1 * 2
```

```python
arr1 / 3
```

```python
a1 = np.arange(9).reshape((3, 3))
a2 = np.arange(9, 18).reshape((3, 3))
a1, a2
```

```python
a1 + a2
```

```python
a1 = np.arange(4)
a2 = np.arange(4, 8)
a1, a2
```

```python
# vector dot product is always a scalar value
np.dot(a1, a2) # 0*4 + 1*5 + 2*6 + 3*7 = 38
```

```python
a1 = np.arange(9).reshape((3, 3))
a2 = np.arange(9, 27, 2).reshape((3, 3))
a1, a2
```

```python
np.dot(a1, a2) # 0*9 + 1*15 + 2*21 = 57, 0*11 + 1*17 + 2*23 = 63, 3*9 + 4*15 + 5*21= 192 .....n
```

```python
a1 = np.arange(8).reshape((2, 2, 2))
a2 = np.arange(1, 3)
print(a1, a2)
print(a1.shape, a2.shape)
```

```python
np.dot(a1, a2) # 0*1 + 1*2 = 2, 2*1 + 3*2 = 6 | 4*1 + 5*2 = 14, 6*1 + 7*2 = 20 -> [2 X 2]
```

```python
a1 = np.arange(18).reshape((3, 3, 2))
a2 = np.arange(8).reshape((2, 4))
print(a1, '\n\n', a2)
print(a1.shape, a2.shape)
```

```python
result = np.dot(a1, a2) 
result
# result.shape = (3, 3, 4)
```

```python
# def cube (x: list):
#     return [ele**3 for ele in x]
# cube([1, 2, 3])

def cube (x: int):
    return x**3

for ele in range(1, 4):
    print(cube(ele))
```

```python
cube_vec = np.vectorize(cube)
cube_vec([1, 2, 3])
```
