# Mathematical & Statistical Operations in NumPy

```python
import numpy as np
```

```python
a1 = np.array([1, 2, 3])
a2 = np.array([4, 5, 6])

np.add(a1, a2)
```

```python
np.subtract(a1, a2)
```

```python
np.multiply(a1, a2)
```

```python
np.divide(a1, a2)
```

```python
np.divmod(a2, a1)
```

```python
np.sum(np.arange(10))
```

```python
a1 = np.random.randint(1, 100) # random values
a1
```

```python
np.random.seed(1) # fixed value for random values
# np.random.random((2, 5))
a1 = np.random.random(10)
a1
```

```python
1e3 # 1e = 10, power=3, for decimal power in minus 
```

```python
# np.random.uniform(10)
np.mean(a1)
```

```python
np.mean(np.random.random((2, 5)))
```

```python
a1 = np.random.random_integers(1, 100, 50)
```

```python
np.max(a1), np.min(a1)
```

```python
np.argmax(a1), np.argmin(a1)
```

```python
a1 = np.random.random_integers(1, 100, (5, 5))
a1
```

```python
np.max(a1), np.min(a1)
```

```python
np.argmax(a1), np.argmin(a1)
#1 d
```

```python
# np.unravel_index(9, a1.shape)
np.unravel_index(np.argmax(a1), a1.shape)
```

```python
np.unravel_index(np.argmin(a1), a1.shape)
```

```python
np.sqrt(a1)
```

```python
np.cbrt(a1)
```

```python
np.log(a1)
```

```python
np.log10(a1)
```
