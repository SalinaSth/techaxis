### Transpose 
- Converts rows into columns and cice versa of matrix (for 2d)

```python
arr = np.arange(1, 5).reshape((2, 2))
arr
```

```python
arr.T
```

```python
np.transpose(arr)
```

```python
# eigen values and eigen vectors
A * e_vector = e_values * e_vector
```

```python
e_val, e_vec = np.linalg.eig(arr)
e_val, e_vec
```

```python
# 2x + 3y = 10
# x - y = 2
```

```python
coeff = np.array([
    [2, 3],
    [1, -1]
])

y = np.array([10, 2])

np.linalg.solve(coeff, y)
```

```python
2*3.2 + 3*1.2
```

```python
# x + y -2z = 7
# -2x + 3y -4z = 5
# 4x - 5y + 6z = 11

coeff = np.array([
    [1, 1, -2],
    [-2, 3, -4],
    [4, -5, 6]
])

y = np.array([7, 5, 11])

np.linalg.solve(coeff, y)
```
