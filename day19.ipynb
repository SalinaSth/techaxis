# DataFrame 
#### tabular structure like excel sheet

```python
pd.DataFrame()
```

```python
# number of elements and values hsould be equal to create data frame
d = {
    "age": [10, 12, 11, 16, 9, 10],
    # "height": [1.5, 1.6, 1.7, 1.8, 1.9]
    "height": [1.5, 1.6, 1.7, 1.8, 1.9, 1.5]
}
df = pd.DataFrame(d)
df
```

```python
# shows value in individual column

l =[
    [10, 12, 11, 16, 9, 10],
    [1.5, 1.6, 1.7, 1.8, 1.9, 1.5]
]
df = pd.DataFrame(l)
df
```

```python
arr = np.array(l).T
df = pd.DataFrame(arr, columns=['age', 'height'])
df
```

```python
l =[
    [10, 1.5],
    [12, 1.6],
    [11, 1.7],
    [16, 1.8],
    [9, 1.9],
    [10, 1.5]
]

df = pd.DataFrame(l, columns = ['age', 'height'])
df
```

```python
# number of elements and values hsould be equal to create data frame
d = {
    "age": [10, 12, 11, 16, 9, 10],
    # "height": [1.5, 1.6, 1.7, 1.8, 1.9]
    "height": [1.5, 1.6, 1.7, 1.8, 1.9, 1.5]
}
df1= pd.DataFrame(d)
df1
```

```python
np.linspace(0, 10, 5)
```

```python
# number of elements and values hsould be equal to create data frame
d = {
    "age": np.arange(50, 60),
    "height": np.linspace(4, 5, 10)
}
df2 = pd.DataFrame(d)
df2
```

```python
pd.concat([df1, df2], axis=0, ignore_index=True) # ignore_space = true, shows continuos index value
```

```python
d1 = {
    'a': np.arange(5),
    'b': np.random.random(5)
}
d1
```

```python
d2 ={'c': np.linspace(10, 19, 5),
     'd': np.linspace(100, 1000, 5)}
d2
```

```python
df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)
```

```python
df1.head(2) # shows 2 records
```

```python
df2.head(3)
```

```python
pd.concat([df1, df2], axis=1, ignore_index=True)
```

```python
df1.columns, df2.columns
```

```python
columns = [col for df in [df1, df2] for col in df.columns]
columns
```

```python
cols = []
for df in [df1, df2]:
    for col in df.columns:
        cols.append()
print(cols)
```

```python
df_cat = pd.concat([df1, df2], axis=1, ignore_index=True)
df_cat
```

```python
df_cat.rename(columns={0: 'age', 1:'height', 2:'place', 3:'grade'})
```

```python
# to update the old dataframe, use either of two below for inplace updation 
# df_cat = df_cat.rename(columns={0: 'age', 1:'height', 2:'place', 3:'grade'})
df_cat.rename(columns={0: 'age', 1:'height', 2:'place', 3:'grade'}, inplace=True)
df_cat
```

```python
df_cat = df_cat.rename(columns={'place': 'roll_num'})
df_cat
```
