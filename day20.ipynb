# Importing data from external source (csv, excel, json)

```python
import pandas as pd
```

```python
# pd.read_ fuctions returning type is dataframe
df = pd.read_csv('./data/data1.csv')
df 
```

- If column name is not provided in csv file, then first record will be treated as column names. To avoid this, we can use pandas read-csv function with header = None and names parameter to specify column names

```python
#if data does not have a column pass header none
df = pd.read_csv('./data/data1_1.csv', header = None)
df 
```

```python

df = pd.read_csv(
    './data/data1_1.csv', 
    header = None, 
    # names = list('abcd') 
    names = ['sn', 'name', 'age', 'height']
)
df 
```

```python
#delimeter: that separates the values by commas or spaces etc
# read_csv default separator is comma
df2 = pd.read_csv('./data/data2.csv', sep= '\t')
df2 
```

## In pandas
- axis=0 : row-wise operation
- axis=1 : column-wise operation

```python
# drop : removes column or wow data with axis 
# axis=1 represent column wise  and axis=0 represents row-wise, its opposite of numpy
df2.drop('Unnamed: 0', axis=1)
```
