### Filtering

```python
data = {
    'Department': ['HR', 'Finance', 'IT', 'HR', 'IT', 'Finance', 'HR'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'Salary': [5000, 6000, 7000, 5200, 7500, 6400, 5500],
    'Experience': [3, 4, 5, 2, 6, 5, 3]
}

df = pd.DataFrame(data)
df
```

```python
# filter method
df.filter(lambda x : x > 6000)
```

```python
# uses broadcasting
condition = df['Salary'] > 6000
condition
```

```python
df[condition]
```

```python
df[df['Salary']>6000] 
```

```python
# df[(condition1) & (condition2) & (condition3)]

df[(df['Salary']>6000) & (df['Department'] == 'IT')] 
```

```python
data = {
    'Department': ['HR', 'Finance', 'IT', 'HR', 'IT', 'Finance', 'HR'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'Salary': [5000, 6000, 7000, 5200, 7500, 6400, 5500],
    'Experience': [3, 4, 5, 2, 6, 5, 3],
    'Distance_Travelled': [10000, 11000, 15000, 100, 1200, 1500, 1750]
}

df = pd.DataFrame(data)
df
```

```python
df[['Salary_thousands', 'Distance_Travelled_in_km']] = df[['Salary', 'Distance_Travelled']]\
    .applymap(lambda x : x /1000)
df
```

```python
df[['Department', 'Employee']] = df[['Department', 'Employee']].applymap(lambda x : x.strip())
df
```
