```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.scatter(np.arange(200), arr, s=10, c='red', marker='D', label='male')
plt.scatter(np.arange(200), arr2, s=20, c='green', marker='*', label='female')
# include information in the plot
plt.legend()
plt.xlabel('earnings in thousands')
plt.ylabel('age')
plt.title('Earnings wrt Age')
# plt.grid(axis='x')
# plt.grid(axis='y')
plt.grid(axis='both') # default
```

```python
# line plot: marker or points
# mec = marker edge color, mfc= face color
# plt.plot(arr)
arr7 = [2, 3, 4, 8, 1, 7]
plt.figure(figsize=(10, 4))
plt.plot(arr7, marker='*', mec='red', mfc='green')
plt.legend()
plt.xlabel('Number of people')
plt.ylabel('income in millions')
plt.title('Earnings wrt Number of people')
plt.grid(axis='both') # default
```

```python
#scatter plot without the marker
plt.plot(arr7, '*', mec='red', mfc='green')
```
