# Subplots & Multiple plots in Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 100, 10)
x
```

```python
np.sin(x), np.cos(x)
```

## .subplot()

```python
plt.figure(figsize=(6, 3))
plt.plot(x)
# plt.plot(np.sin(x))
plt.show()
```

```python
plt.plot(np.sin(x))
```

```python
#2 by 2 has 4 figs, index starts from 1
# plt.titght_layout - prevent from overlapping
# plt.subplot = pass number of rows and number of columns=always same, then index

plt.subplot(2, 2, 1)
plt.scatter(np.arange(len(x)), x)
plt.xlabel('n_x')
plt.ylabel('x')
plt.title('n_x vs x')

plt.subplot(2, 2, 2)
plt.plot(x)
plt.ylabel('x')
plt.title('Line plot')

plt.subplot(2, 2, 3)

plt.subplot(2, 2, 4)
plt.scatter(
    np.arange(len(x)),
    np.exp(x)
)
plt.xlabel('n_x')
plt.ylabel('exp(x)')
plt.title('n_x vs x (exp x)')

plt.suptitle("Age") #supertitle, title of whole plot

plt.tight_layout() # important
```

### .subplots()

```python
# returns figure and axis
fig, ax= plt.subplots(3, 2)

ax[0,0].scatter(np.arange(len(x)),x)
ax[0,0].set_title('Scatter plot')

ax[1, 1].plot(np.sin(x))
ax[1, 1].set_title('Sin plot')

ax[2, 0].hist(np.exp(x))
ax[2, 0].set_title('Exp plot')
ax[2, 0].set_xlabel('x')
ax[2, 0].set_ylabel('exp(x)')

plt.tight_layout()
fig.savefig('temp.png', transparent=False)
```

```python
fig, ax= plt.subplots(3, 2, sharex=True, sharey=True)
# sharex , sharex - use only for uniform plots
# as its compresses so its seen in staright line

ax[0,0].scatter(np.arange(len(x)), np.sin(x))
ax[0,0].set_title('Scatter plot')

ax[0,1].scatter(np.arange(len(x)), np.cos(x))
ax[0,1].set_title('Scatter plot')

ax[1, 1].plot(np.sin(x))
ax[1, 1].set_title('Sin plot')

ax[2, 0].hist(np.exp(x))
ax[2, 0].set_title('Exp plot')
ax[2, 0].set_xlabel('x')
ax[2, 0].set_ylabel('exp(x)')

plt.tight_layout()
fig.savefig('temp.png', transparent=False)
```

```python
import matplotlib.gridspec as gridspec
```

```python
# Create a figure with a grid layout of 2 rows and 2 columns
fig = plt.figure(figsize=(6, 6))
gs = gridspec.GridSpec(2, 2, figure=fig)  # 2x2 grid

# # Create subplots in the grid
ax1 = fig.add_subplot(gs[0, :])  # Top-left , gs[0, :] makes ax1 span both columns in the first row
# ax2 = fig.add_subplot(gs[0, 1])  # Top-right
ax2 = fig.add_subplot(gs[1, 0])  # Bottom-left
ax3 = fig.add_subplot(gs[1, 1])  # Bottom-right

# Set titles for each subplot
ax1.set_title("Wide Plot")
ax2.set_title("Small Left")
ax3.set_title("Small Right")

plt.tight_layout()  # Adjust layout to prevent overlapping
plt.show()
```

```python
fig = plt.figure(figsize=(6, 6))
gs = gridspec.GridSpec(3, 2, height_ratios=[2, 1, 1])  # First row is twice as tall

ax1 = fig.add_subplot(gs[0, :])  # First row (spans both columns)
ax2 = fig.add_subplot(gs[1, 0])  # Middle-left
ax3 = fig.add_subplot(gs[1, 1])  # Middle-right
ax4 = fig.add_subplot(gs[2, :])  # Bottom row (spans both columns)

ax1.set_title("Tall Row")
ax2.set_title("Middle Left")
ax3.set_title("Middle Right")
ax4.set_title("Bottom Row")

plt.tight_layout()
plt.show()
```
