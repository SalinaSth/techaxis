# Data Splitting

```python
import seaborn as sns
```

```python
df = sns.load_dataset("iris")
df
```

```python
# 80% of the data will be used for training and 20% for testing
df.shape # (n_records, n_features)

```

```python
n_records = df.shape[0]
n_train = int(n_records * 0.8)
n_test = n_records - n_train
n_train, n_test, n_train + n_test
```

```python
# first shuffle the data randomly then split it into training and testing sets
df = df.sample(frac = 1)
df['species'].values
# undersampling means that we will randomly select a subset of the data to create a balanced dataset
# downsampling means that creating a smaller dataset by removing some of the data points
```

```python
# unsupervised learning
train = df[:n_train]
test =df[n_train:]
train.shape, test.shape
```

```python
# splitiing independednt and dependent features
X = df.drop('species', axis=1)
y = df['species']
```

```python
# train-test split
X_train = X[:n_train] # slicing the data
y_train = y[:n_train]

X_test = X[n_train:]
y_test = y[n_train:]
```

```python
# train-test split using scikit learn
from sklearn.model_selection import train_test_split
```

```python
# train_test_split(X, y, train_size = 0.8)
# stratify = y, for equal proportion of each class in the training and testing sets
# might not be in balanced proportions if we don't use stratify
X_train, X_test, y_train, y_test =  train_test_split(
    X, y, train_size = 0.8, stratify = y, random_state = 42
) 
# random state for reproducibility
sum(y_train == 'setosa'), sum(y_train == 'versicolor'), sum(y_train == 'virginica')
```
