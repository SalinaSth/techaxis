### Generator

```python
def add(a, b):
    return a + b
add(1, 2)
```

```python
# in generator function, we use yield instead of return
def add(n):
    l = []
    for i in range(1, n):
        l.append(i + 1)
    return l
add(5)
```

```python
def add_gen(n):
    for i in range(1, n):
        yield i + 1
result = add_gen(5)
result
# give generator object as output
```

```python
# using next() to get the next value from the generator object
next(result)
```

### Decorator
- changes the behavior of function

```python
import time

time.time()
```

```python
def add(a, b):
    # time before the loop starts
    start = time.time()
    total = 0
    for i in range(a, b):
        total += i
    # time after the loop ends
    end = time.time()
    duration = end -start # in seconds
    print(f"{duration = }")
    return total

def multiply(a, b):
    start = time.time()
    total = 1
    for i in range(a, b):
        total += i
    end = time.time()
    duration = end -start
    print(f"{duration = }")
    return total

add(1, 5), multiply(2, 5)
```

```python
def duration_decorator(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func()
        end = time.time()
        duration = end - start
        print(f"{duration = }")
        return result
    return wrapper

@duration_decorator # decorator syntax to apply the decorator to the function
def add_dec():
    total = 0
    for i in range(0, 100000):
        total += i
    return total

@duration_decorator
def multiply_dec():
    total = 1
    for i in range(2, 100):
        total *= i
    return total

# add(1, 5), multiply(2, 5)
```
