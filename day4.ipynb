     while condition:

````python
value = 0
steps = 0
while value < 10:
    value += 0.01
    steps += 1
    #print(steps)
    print(f'Iteration{steps}')
````

## For Loop
```python
#Syntax
    for i in iterable:
        #do somethings with it
```

````python
#representation
range(1, 5)
# memory effcient
list(range(1, 5))

range(1, 5), list(range(1, 5))
````

- Open interval (1,5) -> 2, 3, 4
- Closed interval [1,5] -> 1, 2, 3, 4, 5
- Open Closed interval [1,5) -> 1, 2, 3, 4

````python
range(10), range(0, 10)
````

````python
list(range(0, 10, 2))
````

````python
for ele in range(0, 10, 1):
    if ele == 3:
        print('Skipping with continue for value 3')
        continue
    if ele > 6:
        break
    print(ele)
````
