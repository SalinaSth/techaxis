## Overfitting & Regularization Techniques (Discussion)

### Overfitting
- It is the condition when the model memorizes all the training data patterns instead of extracting general pattern in the training data

### Regularization
- used to handle overfitting issue.
- helps to keep the model simple.

#### Types of Regularization
- L1 (Lasso) Regularization
    - adds sum of absolute value of weights.
    - converts some weight values into zero.
    - useful if we hink some features in dataset are irrelevant.
- L2 (Ridge) Regularization
    - adds sum of square of weights to loss as penalty
    - converts some weight values into near zero 
    - useful if we think every features in the dataset has some contribution in output.

- underfitting: if dataset has complex pattern and that ml model cannot capture that pattern, then it is underfitting. training accuracy is not high
- overfitting: when training it shows high accuracy but when testing it has poor performance.
- cannot extract generalization from the training data.
- if training accuracy is high and testing accuracy is low, then it is overfitting.
- if training accuracy is low and testing accuracy is low, then it is underfitting.
