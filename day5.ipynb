````python
#isinstance : 
def add(val1: int, val2: int):
    if not isinstance(val1, int) and not isinstance(val2, int):
        raise TypeError("Type of val1 and val2 are not int")
    total = val1 + val2
    return total

add(1, 2) /1
````

````python
add(1.00, 2.001)
````

````python
add('a', 'b')
````

## Functions-2

````python
def add (val1: int, val2: int = 0): #define defdault value in parameter
    if not isinstance(val1, int) and not isinstance(val2, int):
        raise TypeError("Type of val1 and val2 are not int")
    return val1 + val2
add(1)
````

````python
add(1, 2)
````

````python
#set keyword argument
add(val2=5, val1=3)
````

## Args and Kwargs

# Args

````python
# def addition(*args):

#     print(args)
def addition(val1:int, *args):
    print(type(args))
    print(args)
````

````python
addition(1, 2, 3, 4, 5)
````

# Kwargs

````python
def addition(**kwargs):
    print(type(kwargs))
    print(kwargs)
````

````python
addition(a=1, b=2, c=3)
````

````python
def abc(a, *, val1=None, val2=None): #a=positional arguments, after * send arguments with keyword
    print(a)
````

````python
abc(1, val1=2, val2=3)
````
