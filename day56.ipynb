## Time-Series Analysis - Basics & Theory

### Time Series data
- The sequence of data points collected or recoreded as successive points of time or on time order usually at regualr interval of time. 
- helps to analyze the pattern over time
- helps to forecast future value based on past values.

    #### Examples
    - Daily Stock prices
    - Hourly temperature readingd
    - Monthly electricity consumption
    - Annual sales revenue
    

## Types of Time Series Data 
- Based on Number of variables/features collected
    - *Univariate*: one variable measures over time. for example, the daily closing priuce of a stock, the daily temperature in a city.
    - *Multivariate* : Multiple variables are measured over time. for example, the temperature, humidity, and wind speed ina city over time.
- Based on point of time when data is collected
    - *Discrete time series*: Data is collected at fixed intervals. For example, hourly, daily, weekly, montly.
    - *Continuous Time series*: Data is collected at any time. For example, sensor readings.
- Based on statistical measurement
    - *Stationary Time Series*: The mean and variance of the data do no change over time.
    - *Non-stationary Time Series*: The mean and variance of data change over time

## Components of time series data 
- Trend: The long-term progression of the series 
- Seasonality: Regular fluctuations that occur ai fixed intervals (e.g,. daily, weekly, montly)
- Noise: Random fluctuations around the trend seasonality.

## Terminologies 
1. Lag: It refers to the previous values of time series data thatc an be used to predict the future values. A lag of 1 means the data point from previous time step(t-1) and a lag of 2 means the data point from two previous time steps (t-2).
    - Example: If today is Monday
        - data recorded in Monday os considered as present data (t)
        - data recorded in Sunday is conisdered as past data (t-1) with lag 1
        - data recorded in Saturday is considered as past data (t-2) with lag 2
2. Autocorrelation: It is measurement of similarity between time series and lagged versions of itself over successive time intervals. It helps in identifying repeating patterns or seasonality
3. Stationary: A time series is said o be stationary if its statistical properties such as mean, variance, and autocorrelation do not change over time.
    - How to check for stationary?
        - Visual Inspection
        - Rolling Statistics
        - Augmented dickey Fuller (ADF) test: hypothesis test, null hypothesis is that the series is non-stationary and alternative hypothesis is taht the series is stationary.

## Techniques to make a series stationary 
1. **Differencing**: It is a process of subtracting the previous value from the current value. This helps in removing the trend and making the series stationary.
