# Error Handling

```python
#syntax
try:
    #error prone statement
except:
    # error handling code
finally:
    # cleanup code
```

````python
try:
    result = 1/'0'
except (ZeroDivisionError, TypeError) as e:
#except Exception as e:
    print(e)
    #print(e.__traceback__.tb_lineno)
    result = None
finally:
    print('final block')
````
