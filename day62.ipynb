### Linear Algebra

```python
t4 = tf.constant([[2, 3], [15, 6]])
t5 = tf.constant([[2, 1], [1, 6]])
tf.linalg.matmul(t4, t5)
```

```python
t4 = tf.constant([[2, 3], [15, 6]], dtype=tf.float32)
tf.linalg.inv(t4)
```

```python
tf.linalg.det(t4)
```

## Comparision and Boolean

```python
t6 = tf.constant([1])
t7 = tf.constant([1])

tf.equal(t6, t7)
```

```python
t6 = tf.constant([1])
t7 = tf.constant([1], dtype=tf.float32)

tf.equal(t6, t7)
# have same datatype
```

```python
# !=

tf.not_equal(t6, t7)
```

```python
tf.less(t6, t7)
```
