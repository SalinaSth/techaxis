### List Comprehension

```python
 # calculate squares of numbers in list in new list

l = [2, 3, 3.5, 'a', 4]
# square = list() or []
squared = []
for ele in l:
    # sq = ele * ele
    if isinstance(ele,int | float):
        sq = ele ** 2 # ** exponential operator
    else:
        sq = ele
    squared.append(sq)
squared, ele
```

```python
[ele ** 2 for ele in l ]
```

```python
 # calculate squares of numbers in list in new list

l = [2, 3, 3.5, 'a', 4]
# square = list() or []
squared = []
for ele in l:
    # sq = ele * ele
    if isinstance(ele,int | float):
        sq = ele ** 2 # ** exponential operator
        squared.append(sq)
squared, ele
```

```python
#including if statemnet in list comprehension
[ele ** 2 for ele in l if isinstance(ele, int | float) ]
```

```python
# including if-else statement in list comprehension
# if there is if else statement write it at the upper in for loop
[
    ele**2 if isinstance(ele, int | float) else ele 
    for ele in l
]
```
