```python
set1 = set()
type(set1)
```

```python
# set with default elements, unordered
set2 = {1, 2, 3}
set2
```

```python
#add elements into set, no duplicate value
set1.add(1)
set1
```

```python
set3  = {1, 2, 3, 1, 2, 1, 2, 4, 5, 6, 7, 8}
set3
```

```python
# remove random element from set
popped_value = set3.pop()
popped_value, set3
```

```python
set3.remove(15)
set3
```

```python
#suppress error if element not found
set3.discard(15)
set3
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
```

```python
# union
set1.union(set2)
```

```python
# intersection
set1.intersection(set2)
```

```python
#intersection update : update set1 with intersection of set1 and set2
set1.intersection_update(set2)
set1, set2
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
```

```python
set1.difference(set2)
```

```python
set1.difference_update(set2)
set1, set2
```

```python
set1.update(set2)
set1, set2
```
