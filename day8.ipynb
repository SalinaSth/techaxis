```python
#fromkeys- used to create a new dictionary with specified keys and a single value assigned to all keys.
# initialize multiple keys with same value
# d2.fromkeys(keys, value)

keys = ['name', 'age', 'city']
new_dict = dict.fromkeys(keys, None)
print(new_dict)
```

```python
#setdefault : used to get the value of a specified key in a dictionary
# if the key does not exist, it inserts the key with the specified value.
person = {'name': 'john', 'age': 25}
# value = person.setdefault('age', 30)
value = person.setdefault('city', 'ktm')
print(person)
print(value)
```

## String methods

```python
#casefold() : returns a string where all the characters are lower case.
str1.casefold()
```

```python
#find : returns the index of the first occurrence of the specified value.
str2.find('is')
```

```python
text = 'Hello, world!'
print(text.startswith('Hello'))
print(text.startswith('world'))
```

```python
#replace : used to replace a specified phrase with another specified phrase.
text = 'This is python class'
new_text = text.replace('python', 'data science')
print(new_text)
```

```python
#split : used to split a string into a list where each word is a list item.
data = "onepiece,bleach,jojo"
new_data = data.split(",")
print(new_data)
```

```python
#join : used to join the elements of an iterable to the end of the string.
text = "Hello"
new_text = "-".join(text)
print(new_text)
a = ['Hello', 'world', 'from', 'python']
res = " ".join(a)
print(res)
```

```python
print(text.isalpha())
text2 = "hello112"
print(text2.isalpha())
```

```python
#count : used to count the number of times a specified value appears in the string.
data.count('e')
```
